# The code draws FSC plot on CryoSPARC exported data. Required calculations from spatial frequency to resolution is described in axis_artist() function.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
```

```python
fsc_data = pd.read_csv("example.csv") # exported from cryoSPARC, contains FSC data
```

```python
# map parameters required for plotting

name = "" # name of the map. Leave empty if not required
color= "coral" # Color of the FSC curve. I prefer coral, but you can choose any color you like. You can also use hex codes.

# Essential parameters for plotting. 
# These are required to draw the resolution lines and convert the x-axis from spatial frequency to resolution.
# REPLACE WITH YOUR OWN VALUES!!!
boxsize = 0
pixelsize = 0
resolution = 0
```

```python
# functions of plotting supporting details

ylab = "FSC"
xlab = "Spatial Frequency, "+r"$\AA$"
tick=[i for i in range(0, 350, 50)]

def draw_line(ds, col):
    """
    Draw FSC curve with specified color and linewidth
    """
    plt.plot(ds["wave_number"], ds["fsc_noisesub"], color=col, linewidth=3)

def fsc_resolution_lines(bs, ps, res):
    """
    Draw dotted crossed lines representing resolution
    """

    # 0.143 threshold line
    plt.axhline(y=0.143, color="black", linestyle="--")
    f = ax.text(bs*ps/res-30, 0.175, "CC = 0.143", color="crimson", fontweight="semibold", horizontalalignment='center')
    f.set_bbox(dict(facecolor='white', alpha=0.8, edgecolor='red'))

    # resolution line
    plt.axvline(x=bs*ps/res, color="black", linestyle="--")
    r = plt.text(bs*ps/res, -0.1, f"{res}", color="crimson", fontweight="semibold")
    r.set_bbox(dict(facecolor='white', alpha=0.8, edgecolor='red'))

def axis_artist(bs, ps, name, res):
    """
    Convert axis and write labels
    """

    # Convert x-axis from spatial frequency to resolution
    ax.set_xticks(ticks=tick, labels=[round(bs*ps/i, 1) if i != 0 else 'DC' for i in tick])
    ax.set_ylim(-0.05,1.002)
    ax.set_xlim(0, bs*ps/res+50)

    ax.set(title = "GSFSC, "+name, ylabel=ylab, xlabel=xlab)
```

```python
# Drawing the FSC plot

fig, ax = plt.subplots()
draw_line(fsc_data, color)
fsc_resolution_lines(boxsize, pixelsize, resolution)
axis_artist(boxsize, pixelsize, name, resolution)

if name != "":
    plt.savefig(name+".png")
else:
    plt.savefig("FSC_plot.png")
```
